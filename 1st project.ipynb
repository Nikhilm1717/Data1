```python
import pandas as pd
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/311_Service_Requests_from_2019May.csv')
```

```python
print(df["Complaint Type"].value_counts().head(10))
```

```python
df[df['Complaint Type'].str.contains('Noise')]
```

```python
df2 = df[df['Complaint Type'].str.contains('Noise')]
borough_counts = df2['Borough'].value_counts()
print("Noise Complaints by Borough")
print(borough_counts)
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(9, 5))
sns.barplot(x=borough_counts.index, y=borough_counts.values, palette="Blues_r")
plt.title("Noise Complaints by Borough", fontsize=14, fontweight="bold")
plt.xlabel("Borough", fontsize=11)
plt.ylabel("Number of Noise Complaints", fontsize=11)
plt.ylim(0, borough_counts.max() * 1.15)
plt.tight_layout()

plt.show()
```

```python
df2['Borough'].value_counts().head(10).sort_values(ascending=False).plot(kind='bar')
```

```python
####### Manhattan with most noise complaints followed by Brooklyn !
```
